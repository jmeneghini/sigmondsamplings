```python
# get Ecm
ecm_obs = loader.find_observables(phirho_file_path, "Ecm")
ecm_samplings = loader.load_observables(phirho_file_path, ecm_obs)
ecm_samplings =list(ecm_samplings.values())
ecm_samplings
```

```python
ecm_means = [ecm_samplings[i].mean for i in range(len(ecm_samplings))]
ecm_errs = [ecm_samplings[i].error for i in range(len(ecm_samplings))]

qcm_means = [qcm_samplings[i].mean for i in range(len(qcm_samplings))]
qcm_errs = [qcm_samplings[i].error for i in range(len(qcm_samplings))]
```

```python
# sort box last [index] in name
names = [loader.get_name_and_index_from_dict_key(key)[0] for key in box_samplings.keys()]

get_energy_index = lambda name: int(name[-2:-1] if not name[-3].isdigit() else name[-3:-1])
box_samplings_sorted = [box_samplings[key+"[0]"] for key in sorted(names, key=lambda x: get_energy_index(x))]
# sorted keys
box_samplings_sorted_keys = {key: box_samplings[key+"[0]"] for key in sorted(names, key=lambda x: get_energy_index(x))}
box_samplings_sorted_keys
```

```python
box_means = [box_samplings_sorted[i].mean for i in range(len(box_samplings_sorted))]
box_errs = [box_samplings_sorted[i].error for i in range(len(box_samplings_sorted))]
box_samplings = box_samplings_sorted
```

```python
phase_shift_samplings = []
import numpy as np
# box = qcm/mref cot(delta)
for i in range(len(ecm_samplings)):
    phase_shift_samplings.append(
        np.arctan2(qcm_samplings[i].to_real(),box_samplings[i].to_real())
    )
```

```python
phase_shift_means = [phase_shift_samplings[i].mean for i in range(len(phase_shift_samplings))]
phase_shift_errs = [phase_shift_samplings[i].error for i in range(len(phase_shift_samplings))]
phase_shift_samplings
```

```python
# plot phase shift vs qcm
import matplotlib.pyplot as plt
plt.errorbar(qcm_means, phase_shift_means, xerr=qcm_errs, yerr=phase_shift_errs, fmt='o', label='Phase Shift vs Ecm')
plt.xlabel('qcm')
plt.ylabel('Phase Shift (rad)')
# put pi/2 and pi on the y-axis
plt.yticks([0, np.pi/2, np.pi], ['0', r'$\frac{\pi}{2}$', r'$\pi$'])
plt.savefig('phase_shift_vs_qcm.png')
```

```python

# now plot the samplings
plt.figure(figsize=(10, 6))
for i in range(len(ecm_samplings)):
    plt.scatter(qcm_samplings[i].to_real().resampled_values, phase_shift_samplings[i].to_real().resampled_values, s=10, alpha=0.5, label=f'Ecm_index {i}')
plt.xlabel('qcm')
plt.ylabel('Phase Shift (rad)')
plt.legend()
```
